# Day 1

```python
day1 = []
with open('inputs/day1') as f:
    day1 = [l.strip() for l in f.readlines()]
```

```python
def p1(inpt):
    s = 50
    p = 0
    for i in inpt:
        d, v = i[0], int(i[1:]) % 100
        if d == 'L':
            if s - v < 0:
                s = 100 + (s - v)
            else:
                s -= v
        elif d == 'R':
            if s + v > 99:
                s = (s + v) - 100
            else:
                s += v
        if s == 0:
            p += 1
    return p

p1(day1)
```

```python
def p2(inpt):
    s = 50
    p = 0
    for i in inpt:
        d, v = i[0], int(i[1:])
        p += v // 100
        v %= 100
        if d == 'L':
            if s - v < 0:
                if s != 0:
                    p += 1
                s = 100 + (s - v)
            else:
                s -= v
                if s == 0:
                    p += 1
        elif d == 'R':
            if s + v > 99:
                s = (s + v) - 100
                p += 1
            else:
                s += v
    return p

p2(day1)
```
