# Day 1

```python
day1 = []
with open('inputs/day1') as f:
    day1 = [l.strip() for l in f.readlines()]
```

```python
def p1(inpt):
    s = 50
    p = 0
    for i in inpt:
        d, v = i[0], int(i[1:]) % 100
        if d == 'L':
            if s - v < 0:
                s = 100 + (s - v)
            else:
                s -= v
        elif d == 'R':
            if s + v > 99:
                s = (s + v) - 100
            else:
                s += v
        if s == 0:
            p += 1
    return p

p1(day1)
```

```python
def p2(inpt):
    s = 50
    p = 0
    for i in inpt:
        d, v = i[0], int(i[1:])
        p += v // 100
        v %= 100
        if d == 'L':
            if s - v < 0:
                if s != 0:
                    p += 1
                s = 100 + (s - v)
            else:
                s -= v
                if s == 0:
                    p += 1
        elif d == 'R':
            if s + v > 99:
                s = (s + v) - 100
                p += 1
            else:
                s += v
    return p

p2(day1)
```

# Day 2

```python
day2 = []
with open('inputs/day2') as f:
    day2 = [l.strip() for l in f.readlines()]

day2 = [(int(r.split('-')[0]), int(r.split('-')[1])) for r in day2[0].split(',')]
```

```python
def p1(inpt):
    def is_invalid(num):
        if len(num) % 2 != 0:
            return False
        return num[:len(num) // 2] == num[len(num) // 2:]
    invalid = 0
    for s, e in inpt:
        for n in range(s, e + 1):
            if is_invalid(str(n)):
                invalid += n
    return invalid
p1(day2)
```

```python
def p2(inpt):
    def is_invalid(num):
        for mul in range(1, len(num) // 2 + 1):
            if len(num) % mul == 0:
                if num[:mul] * (len(num) // mul) == num:
                    return True
        return False
    invalid = 0
    for s, e in inpt:
        for n in range(s, e + 1):
            if is_invalid(str(n)):
                invalid += n
    return invalid
p2(day2)
```

```python
from collections import deque

def p2(inpt):
    max_joltage = 0
    for bank in inpt:
        drops = len(bank) - 12
        stack = deque()
        for digit in bank:
            while drops > 0 and stack and stack[-1] < digit:
                stack.pop()
                drops -= 1
            stack.append(digit)
        max_joltage += int("".join(list(stack)[:12]))
        
    return max_joltage
    
p2(day3)
```
