```python
def get_input(f_name):
    if os.path.exists(f'inputs/{f_name}'):
        with open(f'inputs/{f_name}', 'r') as f:
            return [line.strip() for line in f.readlines()]
    else:
        return []
```

```python
def part1(lines):
    left = sorted(list(map(int, [left for left, right in [[nums for nums in line.split()] for line in lines]])))
    right = sorted(list(map(int, [right for left, right in [[nums for nums in line.split()] for line in lines]])))
    total_distance = 0
    for l, r in zip(left, right):
        total_distance += abs(l - r)
    return total_distance
    
part1(inpt)
```

```python
def part2(lines):
    left = [left for left, right in [[nums for nums in line.split()] for line in lines]]
    right = [right for left, right in [[nums for nums in line.split()] for line in lines]]
    hash_map = defaultdict(int)
    total_similarity = 0
    for num in right:
        hash_map[num] += 1
    for num in left:
        total_similarity += int(num) * hash_map[num]
    return total_similarity

part2(inpt)
```

```python
def is_safe(level):
    p_f, p_r = level[0], level[-1]
    asc, desc = True, True
    for f, r in zip(level[1:], level[::-1][1:]):
        if p_f > f or not 1 <= abs(p_f - f) <= 3:
            asc = False
        if p_r > r or not 1 <= abs(p_r - r) <= 3:
            desc = False
        if not asc and not desc:
            break
        p_f = f
        p_r = r
    return asc or desc
```

```python
def part1(lines):
    levels = [list(map(int, line.split())) for line in lines]
    safe_levels = 0
    for level in levels:
        safe_levels += is_safe(level)
    return safe_levels
    
part1(inpt)
```

```python
def part2(lines):
    levels = [list(map(int, line.split())) for line in lines]
    safe_levels = 0
    for level in levels:
        safe = is_safe(level)
        if not safe:
            for i in range(len(level)):
                safe = safe or is_safe(level[:i] + level[i + 1:])
        safe_levels += safe
    return safe_levels

part2(inpt)
```

```python
def part2(lines):
    enabled = True
    total = 0
    for line in lines:
        for match in re.findall(r'(mul\((\d+),(\d+)\)|don\'t\(\)|do\(\))', line):
            if match[0] == "don't()":
                enabled = False
            elif match[0] == 'do()':
                enabled = True
            elif enabled:
                total += int(match[1]) * int(match[2])
    return total

part2(inpt)
```

```python
def part1(lines):
    rules = [line for line in lines if re.search(r'^\d+\|\d+$', line)]
    updates = [line for line in lines if re.search(r'^(\d+|(\d+,)+\d+)$', line.strip())]
    rule_map = defaultdict(lambda: defaultdict(set))
    for rule in rules:
        before, after = rule.split('|')[0], rule.split('|')[-1]
        rule_map[before]['after'].add(after)
        rule_map[after]['before'].add(before)

    good_updates = 0
    for update in updates:
        cur_nums = set()
        nums = update.split(',')
        is_good_update = True
        for i in range(len(nums)):
            if len(rule_map[nums[i]]['after'] & cur_nums) > 0:
                is_good_update = False
                break
            cur_nums.add(nums[i])
        cur_nums = set()
        for i in range(len(nums) - 1, -1, -1):
            if len(rule_map[nums[i]]['before'] & cur_nums) > 0:
                is_good_update = False
                break
            cur_nums.add(nums[i])
        
        if is_good_update:
            good_updates += int(update.split(',')[len(update.split(',')) // 2])
        
    return good_updates

part1(inpt)
```

```python
def perm(target, cur, l, i):
    if cur > target:
        return False
    if i == len(l):
        return cur == target
    else:
        return perm(target, cur * l[i], l, i + 1) or \
               perm(target, cur + l[i], l, i + 1)
```

```python
def part1(lines):
    total = 0
    for line in lines:
        target, nums = line.split(':')
        target = int(target)
        nums = list(map(int, nums.strip().split(' ')))
        total += target * perm(target, nums[0], nums, 1)
    return total
        
part1(inpt)
```

```python
def modified_perm(target, cur, l, i):
    if cur > target:
        return False
    if i == len(l):
        return cur == target
    else:
        return modified_perm(target, cur * l[i], l, i + 1) or \
               modified_perm(target, cur + l[i], l, i + 1) or \
               modified_perm(target, int(f'{cur}{l[i]}'), l, i + 1)
```

```python
def part2(lines):
    total = 0
    for line in lines:
        target, nums = line.split(':')
        target = int(target)
        nums = list(map(int, nums.strip().split(' ')))
        is_valid = modified_perm(target, nums[0], nums, 1)
        total += target * is_valid
    return total    
part2(inpt)
```
