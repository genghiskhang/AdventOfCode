```python
from collections import defaultdict
import re
```

# Day 1

```python
with open('inputs/day1.txt', 'r') as f:
    inpt = [line.strip() for line in f.readlines()]
```

```python
def part1(lines):
    left = sorted(list(map(int, [left for left, right in [[nums for nums in line.split()] for line in lines]])))
    right = sorted(list(map(int, [right for left, right in [[nums for nums in line.split()] for line in lines]])))
    total_distance = 0
    for l, r in zip(left, right):
        total_distance += abs(l - r)
    return total_distance
    
part1(inpt)
```

```python
def part2(lines):
    left = [left for left, right in [[nums for nums in line.split()] for line in lines]]
    right = [right for left, right in [[nums for nums in line.split()] for line in lines]]
    hash_map = defaultdict(int)
    total_similarity = 0
    for num in right:
        hash_map[num] += 1
    for num in left:
        total_similarity += int(num) * hash_map[num]
    return total_similarity

part2(inpt)
```

# Day 2

```python
with open('inputs/day2.txt', 'r') as f:
    inpt = [line.strip() for line in f.readlines()]
```

```python
def is_safe(level):
    p_f, p_r = level[0], level[-1]
    asc, desc = True, True
    for f, r in zip(level[1:], level[::-1][1:]):
        if p_f > f or not 1 <= abs(p_f - f) <= 3:
            asc = False
        if p_r > r or not 1 <= abs(p_r - r) <= 3:
            desc = False
        if not asc and not desc:
            break
        p_f = f
        p_r = r
    return asc or desc
```

```python
def part1(lines):
    levels = [list(map(int, line.split())) for line in lines]
    safe_levels = 0
    for level in levels:
        safe_levels += is_safe(level)
    return safe_levels
    
part1(inpt)
```

```python
def part2(lines):
    levels = [list(map(int, line.split())) for line in lines]
    safe_levels = 0
    for level in levels:
        safe = is_safe(level)
        if not safe:
            for i in range(len(level)):
                safe = safe or is_safe(level[:i] + level[i + 1:])
        safe_levels += safe
    return safe_levels

part2(inpt)
```

# Day 3

```python
with open('inputs/day3.txt', 'r') as f:
    inpt = [line.strip() for line in f.readlines()]
```

```python
def part1(lines):
    return sum(sum(int(first) * int(second) for first, second in re.findall(r'mul\((\d+),(\d+)\)', line)) for line in lines)

part1(inpt)
```

```python
def part2(lines):
    enabled = True
    total = 0
    for line in lines:
        for match in re.findall(r'(mul\((\d+),(\d+)\)|don\'t\(\)|do\(\))', line):
            if match[0] == "don't()":
                enabled = False
            elif match[0] == 'do()':
                enabled = True
            elif enabled:
                total += int(match[1]) * int(match[2])
    return total

part2(inpt)
```
